```python
import random
from collections import defaultdict

# 1. SETUP VARIABLES
STUDENT_LAST_NAME = "PERPO"      
STUDENT_SEED_NUM = 3       
STUDENT_FAVORITE_ARTIST = "RINI" 

# Initialize structures
sources = [f"{STUDENT_LAST_NAME}_T{i}" for i in range(1, 6)]
categories = ['TEMPERATURE', 'VOLTAGE', 'CURRENT', 'PRESSURE', 'SPEED']
random.seed(STUDENT_SEED_NUM)
raw_records = []

# 2. DATA GENERATION
clean_artist = STUDENT_FAVORITE_ARTIST.replace(" ", "").upper()

for idx, char in enumerate(clean_artist):
    source = sources[idx % len(sources)]
    category = categories[(idx + ord(char)) % len(categories)]
    
    base_value = STUDENT_SEED_NUM + (ord(char) * (idx + 1))
    value = round((base_value % 140) + 10.25, 2)
    
    raw_records.append((source, category, value))
    
# Duplicate a record
if raw_records:
    raw_records.append(raw_records[0])
    
if len(raw_records) < 8:
    raw_records.append((sources[0], 'TEMPERATURE', 45.50))
    raw_records.append((sources[1], 'VOLTAGE', 12.60))


# 3. IDENTIFY DISTINCT & REPEATED DATA
seen = set()
duplicates = set()
distinct_records = []

for record in raw_records:
    if record in seen:
        duplicates.add(record)
    else:
        seen.add(record)
        distinct_records.append(record)

# Organize for processing
organized_data = defaultdict(lambda: defaultdict(list))
category_summary = defaultdict(list)

for source, category, value in distinct_records:
    organized_data[source][category].append(value)
    category_summary[category].append(value)

# 4. PRINT OUTPUT REPORTS
print(" ORIGINAL GENERATED DATASET (with duplicates):")
print("-" * 50)
for src, cat, val in raw_records:
    print(f"Source: {src:<12} | Category: {cat:<12} | Value: {val}")

print("\n DUPLICATE ANALYSIS:")
print("-" * 50)
if duplicates:
    for dup in duplicates:
        print(f"Duplicate Found: Source={dup[0]}, Category={dup[1]}, Value={dup[2]}")
else:
    print("No duplicates found.")

print("\n FINAL SUMMARY REPORT BY CATEGORY:")
print("-" * 50)
for cat in sorted(categories):
    vals = category_summary[cat]
    if vals:
        print(f"{cat:<12} -> Samples: {len(vals)} | Min: {min(vals)} | Max: {max(vals)} | Avg: {round(sum(vals)/len(vals), 2)}")
    else:
        print(f"{cat:<12} -> No data samples generated.")
```

```python
import random
from collections import defaultdict


# 1. STUDENT-SPECIFIC INPUTS
LAST_NAME = "PERPO"
SEED_NUM = 3
FAVORITE_ARTIST = "RINI"

# Setup common data based on assignment constraints
customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
products = ['NOTEBOOK', 'PEN', 'USB', 'HEADPHONES', 'CALCULATOR']

# Fixed prices assigned to the five products
product_prices = {
    'NOTEBOOK': 2.50,
    'PEN': 1.20,
    'USB': 15.00,
    'HEADPHONES': 45.00,
    'CALCULATOR': 20.00
}

random.seed(SEED_NUM)
raw_transactions = []
clean_artist = FAVORITE_ARTIST.replace(" ", "").upper()

# 2. TRANSACTION DATA GENERATION
for idx, char in enumerate(clean_artist):
    # Determine customer and product deterministically from character positions
    customer = customers[idx % len(customers)]
    product = products[(idx + ord(char)) % len(products)]
    
    # Generate quantity from SEED_NUM and character positions
    quantity = ((SEED_NUM + ord(char) * (idx + 1)) % 10) + 1
    
    raw_transactions.append((customer, product, quantity))

# RULE: Include at least one repeated customer-product transaction
if raw_transactions:
    repeated_tx = raw_transactions[0]
    raw_transactions.append(repeated_tx)

# 3. IDENTIFY REPEATED & DISTINCT TRANSACTIONS
seen_tx = set()
duplicate_transactions = []
distinct_transactions = []

for tx in raw_transactions:
    if tx in seen_tx:
        duplicate_transactions.append(tx)
    else:
        seen_tx.add(tx)
        distinct_transactions.append(tx)

# 4. DATA PROCESSING & TRANSACTION VALUE
processed_records = []
overall_dataset_value = 0.0

customer_summaries = defaultdict(lambda: {"total_qty": 0, "total_spend": 0.0, "items": set()})
product_summaries = defaultdict(lambda: {"total_qty": 0, "total_revenue": 0.0})

for customer, product, quantity in distinct_transactions:
    price = product_prices[product]
    tx_value = round(quantity * price, 2)
    overall_dataset_value += tx_value
    
    processed_records.append((customer, product, quantity, tx_value))
    
    # Aggregate summaries (Requirement 5)
    customer_summaries[customer]["total_qty"] += quantity
    customer_summaries[customer]["total_spend"] += tx_value
    customer_summaries[customer]["items"].add(product)
    
    product_summaries[product]["total_qty"] += quantity
    product_summaries[product]["total_revenue"] += tx_value

# 5. PRESENT FINAL CONSOLIDATED TRANSACTION REPORT
print("=" * 70)
print("                  CONSOLIDATED TRANSACTION REPORT                 ")
print("=" * 70)

print("\n[1] GENERATED TRANSACTION RECORDS (RAW DATA):")
print(f"{'Customer':<15} | {'Product':<12} | {'Quantity':<8}")
print("-" * 43)
for cust, prod, qty in raw_transactions:
    print(f"{cust:<15} | {prod:<12} | {qty:<8}")

print("\n[2] REPEATED / DISTINCT TRANSACTION ANALYSIS:")
print("-" * 50)
print(f"Total Unique Transactions: {len(distinct_transactions)}")
print(f"Total Repeated Purchases Found: {len(duplicate_transactions)}")
for dup in duplicate_transactions:
    print(f"  ⚠️ Duplicate Purchase detected: {dup[0]} bought {dup[1]} (Qty: {dup[2]}) again.")

print("\n[3] INDIVIDUAL TRANSACTION VALUES (ORDERED BY CUSTOMER):")
print(f"{'Customer':<15} | {'Product':<12} | {'Quantity':<8} | {'Total Value ($)':<15}")
print("-" * 58)
# Sorted in a clean and meaningful order
for cust, prod, qty, val in sorted(processed_records, key=lambda x: (x[0], x[1])):
    print(f"{cust:<15} | {prod:<12} | {qty:<8} | ${val:<14.2f}")

print("\n[4] RELEVANT CUSTOMER PURCHASE SUMMARIES:")
print(f"{'Customer':<15} | {'Unique Products':<15} | {'Total Volume':<12} | {'Total Spent':<12}")
print("-" * 62)
for cust in sorted(customer_summaries.keys()):
    summary = customer_summaries[cust]
    print(f"{cust:<15} | {len(summary['items']):<15} | {summary['total_qty']:<12} | ${summary['total_spend']:<11.2f}")

print("\n[5] RELEVANT PRODUCT SUMMARIES:")
print(f"{'Product':<15} | {'Fixed Price':<12} | {'Total Qty Sold':<15} | {'Total Revenue':<12}")
print("-" * 62)
for prod in sorted(products):
    summary = product_summaries[prod]
    price = product_prices[prod]
    print(f"{prod:<15} | ${price:<11.2f} | {summary['total_qty']:<15} | ${summary['total_revenue']:<11.2f}")

print("\n" + "=" * 70)
print(f" OVERALL DATASET TRANSACTION VALUE:  ${overall_dataset_value:,.2f}")
print("=" * 70)


```

```python
import random


# 0. STUDENT-DETAILS
LAST_NAME = "PERPO"
SEED_NUM = 3
FAVORITE_ARTIST = "RINI"

SUBJECTS = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]
STUDENT_LABELS = ["Alpha", "Beta", "Gamma", "Delta", "Epsilon"]


# 1. HELPERS & SCORE GENERATION RULE
def get_performance_category(score):
    """Determines the performance category based on a standard grading scale."""
    if score is None:
        return "Incomplete"
    elif score >= 90:
        return "Excellent"
    elif score >= 80:
        return "Very Good"
    elif score >= 70:
        return "Good"
    elif score >= 60:
        return "Pass"
    else:
        return "Fail"

def generate_score(student_idx, subj_idx, seed, artist):
    """Generates a deterministic score between 50 and 100 using seed and artist characters."""
    char_val = ord(artist[subj_idx % len(artist)])
    # A pseudo-random combination formula to ensure unique, varied scores
    score = (seed + char_val * (student_idx + 1) + (subj_idx * 7)) % 51 + 50
    return score


# 2. DATA GENERATION
records = []

# Generate standard records
for s_idx, label in enumerate(STUDENT_LABELS):
    student_name = f"{LAST_NAME}_{label}"
    for sub_idx, subject in enumerate(SUBJECTS):
        score = generate_score(s_idx, sub_idx, SEED_NUM, FAVORITE_ARTIST)
        records.append({
            "Student": student_name,
            "Subject": subject,
            "Score": score
        })

# Include exactly one repeated student-subject record
if records:
    repeated_base = records[0]
    records.append({
        "Student": repeated_base["Student"],
        "Subject": repeated_base["Subject"],
        "Score": repeated_base["Score"]
    })

# Include exactly one student record with a missing score (Incomplete)
records.append({
    "Student": f"{LAST_NAME}_{STUDENT_LABELS[-1]}",
    "Subject": SUBJECTS[-1],
    "Score": None  # Missing score
})


# 3. DATA PROCESSING & ANALYSIS
valid_records = []
repeated_records = []
incomplete_records = []
seen_pairs = set()

# Separate and clean the data
for record in records:
    student = record["Student"]
    subject = record["Subject"]
    score = record["Score"]
    pair = (student, subject)
    
    if score is None:
        incomplete_records.append(record)
    elif pair in seen_pairs:
        repeated_records.append(record)
    else:
        seen_pairs.add(pair)
        # Determine category for valid data
        record["Category"] = get_performance_category(score)
        valid_records.append(record)

# Sort valid results by Student name, then by Subject for a clean structure
valid_records.sort(key=lambda x: (x["Student"], x["Subject"]))

# Calculate Summaries
student_summary = {}
subject_summary = {sub: {"total": 0, "count": 0} for sub in SUBJECTS}

for rec in valid_records:
    stu = rec["Student"]
    sub = rec["Subject"]
    scr = rec["Score"]
    
    # Student accumulation
    if stu not in student_summary:
        student_summary[stu] = {"total": 0, "count": 0}
    student_summary[stu]["total"] += scr
    student_summary[stu]["count"] += 1
    
    # Subject accumulation
    subject_summary[sub]["total"] += scr
    subject_summary[sub]["count"] += 1

# 4. FINAL CONSOLIDATED REPORT 
print("=" * 65)
print("              FINAL STUDENT PERFORMANCE REPORT              ")
print("=" * 65)

print("\n[1] VALID PERFORMANCE RECORDS:")
print(f"{'Student Name':<20} | {'Subject':<15} | {'Score':<6} | {'Category':<12}")
print("-" * 65)
for rec in valid_records:
    print(f"{rec['Student']:<20} | {rec['Subject']:<15} | {rec['Score']:<6} | {rec['Category']:<12}")

print("\n[2] ANOMALIES DETECTION (REPEATED & INCOMPLETE RECORDS):")
print("-" * 65)
print("Repeated Records Found:")
if repeated_records:
    for rec in repeated_records:
        print(f"  • {rec['Student']} - {rec['Subject']} (Score: {rec['Score']})")
else:
    print("  • None")

print("\nIncomplete Records Found:")
if incomplete_records:
    for rec in incomplete_records:
        print(f"  • {rec['Student']} - {rec['Subject']} (Score: MISSING)")
else:
    print("  • None")

print("\n[3] PERFORMANCE SUMMARIES:")
print("-" * 65)
print("Student Averages:")
for stu, data in student_summary.items():
    avg = data["total"] / data["count"]
    print(f"  • {stu:<20} : Average Score = {avg:.2f}")

print("\nSubject Averages:")
for sub, data in subject_summary.items():
    if data["count"] > 0:
        avg = data["total"] / data["count"]
        print(f"  • {sub:<15} : Average Score = {avg:.2f}")
    else:
        print(f"  • {sub:<15} : No data")
print("=" * 65)
```
